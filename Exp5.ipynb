```python
func = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(func(5))
print(func(-3))
print(func(0))
```

```python
func = [lambda arg=x: arg * 10 for x in range(1, 5)]
for i in func:
    print(i())
```

```python
s = {"a", "b", "c"}
s.add("d")
s.remove("c")
print(s)
```

```python
a = {"x", "y"}
b = {"y", "z"}
u = a.union(b)
print(a|b)
print(u)
```
