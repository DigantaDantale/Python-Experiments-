```python
# Implicit Type Conversion
x = 5
print("x is of type:", type(x))

y = 3.4
print("y is of type:", type(y))

x = x + y
print(x)
print("x is of type:", type(x))

# Explicit Type Conversion
s = "1010"

e = float(s)
print("After converting to float :", e)

c = int(s, 2)
print("After converting to integer base 2 :", c)

s = "a"
c = ord(s)
print("After converting character to integer :", c)

a = chr(99)
b = chr(67)
print(a)
print(b)

c = hex(10)
print("After converting 10 to hexadecimal string :", c)

c = oct(10)
print("After converting 10 to octal string :", c)

s = 'mit'
print("Tuple:", tuple(s))
print("Set:", set(s))
print("List:", list(s))
```

```python
def zero(n):
    print(id(n))
    n = 0
    print(id(n))

a = 100
print(id(a))
zero(a)
print(a)

def add(lst):
    print(id(lst))
    lst.append(0)
    print(id(lst))

l = [1, 2, 3]
print(id(l))
add(l)
print(l)
```

```python
def check_string(my_string):
    upper = 0
    lower = 0

    for ch in my_string:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1

    print("String:", my_string)
    print("Uppercase:", upper)
    print("Lowercase:", lower)

user_input = input("Enter string: ")
check_string(user_input)
```

```python
import random

def generate_random_number(low, high):
    print("Random float:", random.uniform(float(low), float(high)))
    print("Random int:", random.randint(int(low), int(high)))
    print("Random [0,1]:", random.random())

low = input("Enter lower range: ")
high = input("Enter higher range: ")

generate_random_number(low, high)
```

```python
# Keyword arguments
def printinfo(name, location):
    print("Name:", name)
    print("Location:", location)

printinfo(location='kothrud', name="MIT")

# Default arguments
def printinfo(name, location='kothrud'):
    print("Name:", name)
    print("Location:", location)

printinfo(location='karveroad', name="SNDT")
printinfo(name="MIT")

# Variable length arguments
def printinfo(arg1, *n):
    print("Output is:", arg1)
    for i in n:
        print(i)

printinfo(10)
printinfo(70, 60, 50, 40, 30)
```
